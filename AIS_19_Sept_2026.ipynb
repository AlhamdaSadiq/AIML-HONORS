```python
df_weather = pd.DataFrame({'City': ['New York', 'Los Angeles', 'Chicago'],
                           'Temperature': [30, 25, 20],
                           'Humidity': [70, 60, 80], 'Date': ['2023-01-01', '2023-01-02', '2023-01-03']})
df_weather
```

```python
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'], 
                   'Age': [4, 5, 6],
                    'Salary': [50000, 60000, None],
  'Department': [1, 2, 3],
  'Manager':[None, 'John', None],
   'skills': [['Python', 'SQL'], ['Java', 'C++'], None]})

df_new = df[(df['Age'] > 4) & (df['Salary'] < 70000)]
df_new
```

```python
df.isna()
```

```python
df.fillna({'Manager': 'Unknown'}, inplace=True)
```

```python
df.fillna({'Salary': df['Salary'].mean()}, inplace=True)
```

```python
df.dropna(subset=['Salary'], inplace=True)
df
```

```python
df.dropna()
```

```python
department_df = pd.DataFrame({"id": [1, 2, 3, 4],
                              "department_name": ["HR", "IT", "Finance", "Technical"]})
```

```python
df_merged = pd.merge(df, department_df, 
                     left_on='Department', 
                     right_on='id', how='outer')
df_merged
```

```python
df_additional = pd.DataFrame({'Name': ['ABC', 'PQR', 'XYZ'], 
                   'Age': [14, 15, 16],
                    'Salary': [None, 60000, 40000],
  'Department': [4, 2, 1],
  'Manager':['Alice', 'Bob', 'Charlie'],
   'skills': [['Python', 'SQL'], ['Java', 'C++'], None]})
```

```python
df_new = pd.concat([df, df_additional], ignore_index=True)
df_new
```

```python
new_df=df.copy()
```

```python
new_df = df_additional.groupby('Department')['Salary'].mean().reset_index()
```

```python
new_df.rename(columns={'Salary': 'Average_Salary'}, inplace=True)
```

```python
new_df.dropna()
```

```python
new_df['Average_Salary'] = new_df['Average_Salary'].fillna(
    new_df['Average_Salary'].mean()
)
```

```python
print(new_df)
```

```python
new_df.filter(new_df['Average_Salary']> 500000.0)
```

```python
new_df.pipesort_values(by='Average_Salary',ascending=False)
new_df.pipe(
    lambda x: x.sort_values(by='Average_Salary', ascending=False)
)
```
